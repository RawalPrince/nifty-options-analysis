```python
import requests
import pandas as pd
import zipfile
import io
import os
from datetime import date, timedelta
import time

# ── Config ────────────────────────────────────────────────
SAVE_PATH  = "../data/raw/"
START_DATE = date(2024, 12, 1)
END_DATE   = date(2025, 5, 30)

headers = {
    "User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10_15_7)",
    "Accept-Language": "en-US,en;q=0.9",
    "Accept-Encoding": "gzip, deflate, br",
    "Referer": "https://www.nseindia.com/"
}

# ── Session ───────────────────────────────────────────────
session = requests.Session()
session.get("https://www.nseindia.com", headers=headers)

# ── Downloader ────────────────────────────────────────────
def download_bhavcopy(dt, save_path, session):
    date_str = dt.strftime("%Y%m%d")
    filename = f"BhavCopy_NSE_FO_0_0_0_{date_str}_F_0000.csv.zip"
    url      = f"https://nsearchives.nseindia.com/content/fo/{filename}"
    save_file = os.path.join(save_path, f"{date_str}.csv")

    # Skip if already downloaded
    if os.path.exists(save_file):
        print(f"⏭️  Already exists: {dt}")
        return True

    try:
        r = session.get(url, headers=headers, timeout=15)
        if r.status_code != 200:
            print(f"❌ {dt} — HTTP {r.status_code}")
            return False

        zf  = zipfile.ZipFile(io.BytesIO(r.content))
        df  = pd.read_csv(zf.open(zf.namelist()[0]))

        # Filter NIFTY options only right away
        nifty = df[
            (df["TckrSymb"] == "NIFTY") &
            (df["FinInstrmTp"] == "IO")
        ].copy()

        if nifty.empty:
            print(f"⚠️  No NIFTY data: {dt}")
            return False

        nifty.to_csv(save_file, index=False)
        print(f"✅ {dt} — {len(nifty)} rows")
        return True

    except Exception as e:
        print(f"❌ {dt} — {e}")
        return False

# ── Run ───────────────────────────────────────────────────
current = START_DATE
failed  = []

while current <= END_DATE:
    if current.weekday() < 5:  # skip weekends
        success = download_bhavcopy(current, SAVE_PATH, session)
        if not success:
            failed.append(current)
        time.sleep(0.5)  # be polite to NSE server
    current += timedelta(days=1)

print(f"\n✅ Done. Failed: {len(failed)} days")
if failed:
    print("Failed dates:", failed)
```

```python
from datetime import date, timedelta
import time, os, zipfile, io, requests, pandas as pd

SAVE_PATH  = "../data/raw/"
START_DATE = date(2024, 12, 1)
END_DATE   = date(2025, 5, 30)

headers = {
    "User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10_15_7)",
    "Accept-Language": "en-US,en;q=0.9",
    "Accept-Encoding": "gzip, deflate, br",
    "Referer": "https://www.nseindia.com/"
}

session = requests.Session()
session.get("https://www.nseindia.com", headers=headers)

def download_bhavcopy(dt, save_path, session):
    date_str  = dt.strftime("%Y%m%d")
    save_file = os.path.join(save_path, f"{date_str}.csv")

    if os.path.exists(save_file):
        print(f"⏭️  Already exists: {dt}")
        return True

    url = f"https://nsearchives.nseindia.com/content/fo/BhavCopy_NSE_FO_0_0_0_{date_str}_F_0000.csv.zip"

    try:
        r = session.get(url, headers=headers, timeout=15)
        if r.status_code != 200:
            print(f"❌ {dt} — HTTP {r.status_code}")
            return False

        zf    = zipfile.ZipFile(io.BytesIO(r.content))
        df    = pd.read_csv(zf.open(zf.namelist()[0]))

        # ✅ Correct filter now
        nifty = df[
            (df["TckrSymb"]    == "NIFTY") &
            (df["FinInstrmTp"] == "IDO")        # options only
        ].copy()

        if nifty.empty:
            print(f"⚠️  No NIFTY data: {dt}")
            return False

        nifty.to_csv(save_file, index=False)
        print(f"✅ {dt} — {len(nifty)} rows")
        return True

    except Exception as e:
        print(f"❌ {dt} — {e}")
        return False

# ── Run ───────────────────────────────────────────────────
current = START_DATE
failed  = []

while current <= END_DATE:
    if current.weekday() < 5:
        success = download_bhavcopy(current, SAVE_PATH, session)
        if not success:
            failed.append(current)
        time.sleep(0.5)
    current += timedelta(days=1)

print(f"\n✅ Done. Failed: {len(failed)} days")
```

```python
import pandas as pd
import os

SAVE_PATH = "../data/raw/"
all_files = sorted([f for f in os.listdir(SAVE_PATH) if f.endswith(".csv")])

dfs = []
for file in all_files:
    df = pd.read_csv(os.path.join(SAVE_PATH, file))
    dfs.append(df)

combined = pd.concat(dfs, ignore_index=True)

# ── Rename to readable columns ─────────────────────────
combined = combined.rename(columns={
    "TradDt":          "date",
    "TckrSymb":        "symbol",
    "XpryDt":          "expiry",
    "StrkPric":        "strike",
    "OptnTp":          "option_type",
    "OpnPric":         "open",
    "HghPric":         "high",
    "LwPric":          "low",
    "ClsPric":         "close",
    "SttlmPric":       "settle_price",
    "OpnIntrst":       "open_interest",
    "ChngInOpnIntrst": "oi_change",
    "TtlTradgVol":     "volume",
    "TtlTrfVal":       "turnover",
    "UndrlygPric":     "underlying_price"
})

# ── Fix data types ─────────────────────────────────────
combined["date"]   = pd.to_datetime(combined["date"])
combined["expiry"] = pd.to_datetime(combined["expiry"])
combined["strike"] = combined["strike"].astype(float)
combined["close"]  = combined["close"].astype(float)
combined["open_interest"] = combined["open_interest"].astype(float)
combined["oi_change"]     = combined["oi_change"].astype(float)
combined["volume"]        = combined["volume"].astype(float)
combined["underlying_price"] = combined["underlying_price"].astype(float)

# ── Keep only useful columns ───────────────────────────
cols = ["date", "symbol", "expiry", "strike", "option_type",
        "open", "high", "low", "close", "settle_price",
        "underlying_price", "open_interest", "oi_change", "volume"]

combined = combined[cols]

# ── Save ───────────────────────────────────────────────
combined.to_csv("../data/processed/nifty_options_clean.csv", index=False)

print(f"✅ Total records : {len(combined):,}")
print(f"✅ Date range    : {combined['date'].min().date()} → {combined['date'].max().date()}")
print(f"✅ Unique expiries: {combined['expiry'].nunique()}")
print(f"✅ Option types  : {combined['option_type'].unique()}")
print(f"✅ Shape         : {combined.shape}")
combined.head()
```
